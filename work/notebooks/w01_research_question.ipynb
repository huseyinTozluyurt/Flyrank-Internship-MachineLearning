## 2. The question: decision, action, cost of a wrong call

**Research Question:** To what extent does the time elapsed since a page's last update interact with its position volatility and impression volume to predict search traffic decay, and can an interpretable decision tree isolate actionable refresh thresholds?

*   **Decision:** Which specific pages require an editorial refresh or content update in the upcoming planning cycle?
*   **Action:** Allocate editorial resources exclusively to high-visibility, declining pages identified by the model.
*   **Cost of a wrong call:** False positives waste valuable writer hours updating stable pages that didn't need changes; false negatives allow high-traffic pages to silently decay and lose search market share.

```python
# Quick check on the proportion of declining pages
declining_rate = (df['trend_direction'].str.lower() == 'down').mean()
print(f"Overall dataset declining rate: {declining_rate:.3f}")
```
